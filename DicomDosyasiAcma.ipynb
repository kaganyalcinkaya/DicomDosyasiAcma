# Dicom dosyası açma

```python
import pydicom
import matplotlib.pyplot as plt

# Dosya yolunu belirtin
dosya_yolu = r'C:\dicom.dcm'

# DICOM dosyasını okur
ds = pydicom.dcmread(dosya_yolu)

# Görüntü verisini (pixel_array) alır
goruntu = ds.pixel_array

# Görselleştir
plt.figure(figsize=(10, 10))
plt.imshow(goruntu, cmap=plt.cm.bone) #
plt.axis('off')
plt.show()
```
